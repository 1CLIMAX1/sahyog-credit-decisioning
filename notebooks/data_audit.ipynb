```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Coding Folders\hilbit_ai_assignment\data\loan_applications.csv")

# ── Basic shape ──────────────────────────────────────────────
print(df.shape)
print(df.dtypes)
print(df.isnull().sum())

# ── Numeric summary ──────────────────────────────────────────
print(df.describe())

# ── Categorical columns — check unique values ────────────────
cat_cols = df.select_dtypes("object").columns
for col in cat_cols:
    print(f"\n{col}: {df[col].unique()}")

# ── Suspicious values to look for ───────────────────────────
print(df[df["age"] < 18])               # underage applicants
print(df[df["age"] > 80])               # unrealistically old
print(df[df["monthly_income"] <= 0])    # zero or negative income
print(df[df["ltv_pct"] > 100])          # LTV above 100% (impossible)
print(df[df["foir_pct"] > 100])         # FOIR above 100%
print(df[df["loan_amount"] <= 0])       # zero loan amount
print(df[df["bureau_score"] == -1].shape[0])  # new-to-credit count
print(df[df["interest_rate_pct"] <= 0]) # zero interest rate
```
